```python
# Importação das bibliotecas necessárias e carregamento do dataset
import pandas as pd
from IPython.display import display

dados_ibovespa = pd.read_csv('dados_historicos_ibovespa.csv')
```

```python
# Inspeção do dataset
print("Dados Iníciais do DataFrame:")
display(dados_ibovespa.head(5)) # As Colunas Data, Vol. e Var% são str deveriam ser object

print("Dados Finais do DataFrame:")
display(dados_ibovespa.tail(5)) # As Colunas Data, Vol. e Var% são str deveriam ser object

print("Quandidade de linhas e colunas:")
print(f"Linhas: {dados_ibovespa.shape[0]}")
print(f"Colunas: {dados_ibovespa.shape[1]}")
```

```python
# Tipos dos dados
print("Informações gerais do DataFrame:")
display(dados_ibovespa.info())
```

Ponto de atenção

O `Dtype` das colunas `Data`, `Vol.` e `Var%` estão com tipo `str`, mas deveriam ser `object`

```python
# Estatística descritiva

print("Descrição estatística do DataFrame:")
display(dados_ibovespa.describe())
```

```python
# Missing values

print("Valores ausentes: \n")
missing = dados_ibovespa.isnull().sum()
display(missing)
```

```python
# Duplicidade

print("Valores Duplicados:")
duplicated = dados_ibovespa.duplicated().sum()
print(duplicated)

datas_duplicadas = dados_ibovespa['Data'].duplicated().sum()
print(f"Quantidade de datas duplicadas: {datas_duplicadas}")
```

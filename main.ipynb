## Etapa 7: Verificação de Inconsistências

### Regras de Negócio
- No mercado financeiro a `Máxima` de um dia nunca pode ser menor que a `Mínima` ou que a `Abertura` / `Último`.
- A `Mínima` nunca pode ser maior que qualquer outra cotação do dia.

### Coerência Temporal / Sequencial
- As datas estão ordenadas cronologicamente?
- Existem "saltos" anormais ou datas futuras?

### Formatos/Tipos
- Colunas numericas que contêm caracteres especias que impedem cálculos (como sufixo `M` em `Vol.`, vírgula `,` como separador decimal, `%` na variação)

```python
inconsistencias_max_min = dados_ibovespa[dados_ibovespa['Máxima'] < dados_ibovespa['Mínima']]
inconsistencias_max_abertura = dados_ibovespa[dados_ibovespa['Máxima'] < dados_ibovespa['Abertura']]
inconsistencias_min_abertura = dados_ibovespa[dados_ibovespa['Mínima'] > dados_ibovespa['Abertura']]
inconsistencias_max_ultimo = dados_ibovespa[dados_ibovespa['Máxima'] < dados_ibovespa['Último']]
inconsistencias_min_ultimo = dados_ibovespa[dados_ibovespa['Mínima'] > dados_ibovespa['Último']]

print(f"Máxima menor que Mínima: {len(inconsistencias_max_min)}")
print(f"Máxima menor que Abertura: {len(inconsistencias_max_abertura)}")
print(f"Mínima maior que Abertura: {len(inconsistencias_min_abertura)}")
print(f"Máxima menor que Último: {len(inconsistencias_max_ultimo)}")
print(f"Mínima maior que Último: {len(inconsistencias_min_ultimo)}")

# Verificar se há valores negativos ou zerados onde não deveria
precos_negativos = dados_ibovespa[(dados_ibovespa[['Último', 'Abertura', 'Máxima', 'Mínima']] <= 0).any(axis=1)]
print(f"Linhas com preços <= 0: {len(precos_negativos)}")
```

### Conversão

```python
df_limpo = dados_ibovespa.copy()

df_limpo['Data'] = pd.to_datetime(df_limpo['Data'], format='%d.%m.%Y')

df_limpo['Var%'] = (
    df_limpo['Var%']
    .str.replace('%', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)

def converter_volume (valor):
    if pd.isna(valor) or valor == '-' or valor == "":
        return 0.0
    valor = str(valor).replace('.', '').replace(',', '.')
    if 'M' in valor:
        return float(valor.replace('M', '')) * 1e6
    elif 'B' in valor:
        return float(valor.replace('B', '')) * 1e9
    elif 'K' in valor:
        return float(valor.replace('K', '')) * 1e3
    else:
        return float(valor)

df_limpo['Vol.'] = df_limpo['Vol.'].apply(converter_volume)

df_limpo = df_limpo.sort_values(by='Data').reset_index(drop=True)

display(df_limpo.head(20))
df_limpo.info()
```

### Padronização
